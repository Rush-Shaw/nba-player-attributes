#### Both Attribute Data and Stats Data will be Merged

##### Imports

```python
from pathlib import Path
import json
import pandas as pd
```

##### Directories and File Locations

```python
BASE_DIR = Path.cwd().parent
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
stats_df = pd.read_json(PROCESSED_DIR / "all_nba_stats.json")

attributes_df = pd.read_csv(PROCESSED_DIR / "all_attributes.csv")
```

##### Validate Headings

```python
print(f"Stats DataFrame shape: {stats_df.columns}")
```

```python
print(f"Attributes DataFrame shape: {attributes_df.columns}")
```

##### Rename Stats Column Header for Name

```python
stats_df = stats_df.rename(columns = {
    "PLAYER_NAME": "name"
})
```

##### Merge Datasets

```python
merged_df = stats_df.merge(
    attributes_df,
    on=["name", "season"],
    how="inner"
)
```

##### Validate the merge

```python
missing = merged_df.isnull().sum()

print(missing[missing > 0])
```

Drop Team and that one attribute

```python
merged_df = merged_df.drop(columns=["team"])
```

```python
merged_df = merged_df.dropna()
```

Save Completed Dataset

```python
merged_df.to_csv("nba_ml_dataset.csv", index=False)
```
