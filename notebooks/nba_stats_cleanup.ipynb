#### This notebook will be to process and clean the nba stats data

##### Imports

```python
import json
from pathlib import Path
```

##### Load in Raw Data

```python
project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

with open(project_root / "data/raw/all_nba_stats.json", "r") as f:
    data = json.load(f)
```

```python
print(json.dumps(data, indent=4))
```

##### Remove Uncessary Keys

```python
# All the keys to be removed from the stats data
keys_to_remove = ['PLAYER_ID', 'NICKNAME_base', "TEAM_ID",
                     "AGE_base", "NBA_FANTASY_PTS", "DD2",
                     "TD3", "WNBA_FANTASY_PTS", "GP_RANK_base",
                     "W_RANK_base", "L_RANK_base", "W_PCT_RANK_base",
                     "MIN_RANK_base", "FGM_RANK_base", "FGA_RANK_base",
                     "FG_PCT_RANK_base", "FG_PCT_RANK_base",
                     "FG3M_RANK_base", "FG3A_RANK", "FG3_PCT_RANK", 
                     "FTM_RANK", "FTA_RANK", "FT_PCT_RANK", "OREB_RANK",
                     "DREB_RANK", "REB_RANK", "AST_RANK", "TOV_RANK",
                     "STL_RANK", "BLK_RANK", "BLKA_RANK", "PF_RANK", 
                     "PFD_RANK", "PTS_RANK", "PLUS_MINUS_RANK", 
                     "NBA_FANTASY_PTS_RANK", "DD2_RANK", "TD3_RANK", 
                     "WNBA_FANTASY_PTS_RANK", "TEAM_COUNT_base", 
                     "NICKNAME_advanced", "AGE_advanced", "GP_advanced", 
                     "W_advanced", "L_advanced", "W_PCT_advanced", 
                     "MIN_advanced", "FG_PCT_advanced", "GP_RANK_advanced", 
                     "W_RANK_advanced", "L_RANK_advanced", "W_PCT_RANK_advanced", 
                     "MIN_RANK_advanced", "E_OFF_RATING_RANK", "OFF_RATING_RANK", 
                     "sp_work_OFF_RATING_RANK", "E_DEF_RATING_RANK", 
                     "DEF_RATING_RANK", "sp_work_DEF_RATING_RANK", "E_NET_RATING_RANK", 
                     "NET_RATING_RANK", "sp_work_NET_RATING_RANK", "AST_PCT_RANK", 
                     "AST_TO_RANK", "AST_RATIO_RANK", "OREB_PCT_RANK", "DREB_PCT_RANK", 
                     "REB_PCT_RANK", "TM_TOV_PCT_RANK", "E_TOV_PCT_RANK", "EFG_PCT_RANK", 
                     "TS_PCT_RANK", "USG_PCT_RANK", "E_USG_PCT_RANK", "E_PACE_RANK", 
                     "PACE_RANK", "sp_work_PACE_RANK", "PIE_RANK", "FGM_RANK_advanced", 
                     "FGA_RANK_advanced", "FGM_PG_RANK", "FGA_PG_RANK", "FG_PCT_RANK_advanced", 
                     "TEAM_COUNT_advanced"]
```

```python
for player in data:
    for key in keys_to_remove:
        player.pop(key, None)  # Remove the key if it exists, do nothing otherwise
```

##### Update Season to only show the final year instead of the split

eg) 2021-22 will be changed to 2022

```python
for player in data:
    if player["season"] == "2021-22":
        player["season"] = "2022"
    if player["season"] == "2022-23":
        player["season"] = "2023"
    if player["season"] == "2023-24":
        player["season"] = "2024"
    if player["season"] == "2025-26":
        player["season"] = "2026"
```

##### Dump Data into Processed JSON File

```python
with open(project_root / "data/processed/all_nba_stats.json", "w") as f:
    json.dump(data, f, indent=4)
```
