#### Multi-Year Attributes will be Combined Here

##### Imports

```python
from pathlib import Path
import json
import pandas as pd
```

##### Directories and File Locations

```python
from pathlib import Path

BASE_DIR = Path.cwd().parent
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
attribute_files = {
    2022: RAW_DIR / "2k22_roster.json",
    2023: RAW_DIR / "2k23_roster.json",
    2024: RAW_DIR / "2k24_roster.json",
    2026: PROCESSED_DIR / "2k26_roster.json"
}
```

##### Combine Seasons

```python
all_players = []

for season, file_path in attribute_files.items():
    with open(file_path, "r") as f:
        players = json.load(f)

    for player in players:
        player["season"] = season
        all_players.append(player)

# Create DataFrame from combined player attributes
attributes_df = pd.DataFrame(all_players)
attributes_df.head() # Display the first few rows of the combined DataFrame
```

##### Check Shape

```python
# attributes_df.shape
# attributes_df.columns.tolist()
attributes_df.isnull().sum().sort_values(ascending=False).head(5)
```

##### Unify Agility
###### Feature Harmonization

```python
attributes_df["agility"] = attributes_df["agility"].fillna(
    (
        attributes_df["acceleration"] +
        attributes_df["lateralQuickness"]
    ) / 2
)
```

##### Remove Old Movement Columns

```python
attributes_df.drop(
    columns=[
        "acceleration",
        "lateralQuickness"
    ],
    inplace=True
)
```

Save all roster data from all years into a csv file

```python
attributes_df.to_csv(
    "../data/processed/all_attributes.csv",
    index=False
)
```
